```python
from ultralytics import YOLO
import os

image_path = '/home/dll1305/Documents/ai-project/tf_models/object_detection/quantization/AI_Project/test/munich/'
image_name = "munich_000000_000019_leftImg8bit.png"
```

```python
model_pt = YOLO('/home/dll1305/Documents/ai-project/tf_models/object_detection/quantization/AI_Project/train15/weights/best.pt', task="detect")

output_dir_pt = '/home/dll1305/Documents/ai-project/tf_models/object_detection/quantization/AI_Project/output_pt'
os.makedirs(output_dir_pt, exist_ok=True)
results = model_pt(image_path + image_name)
output_path_pt = os.path.join(output_dir_pt, image_name)
results[0].save(filename=output_path_pt)

print(f"Inference result saved to: {output_path_pt}")
```

```python
model_onnx = YOLO('/home/dll1305/Documents/ai-project/tf_models/object_detection/quantization/AI_Project/train15/weights/best.onnx', task="detect")

output_dir_onnx = '/home/dll1305/Documents/ai-project/tf_models/object_detection/quantization/AI_Project/output_onnx'
os.makedirs(output_dir_onnx, exist_ok=True)
results = model_onnx(image_path + image_name)
output_path_onnx = os.path.join(output_dir_onnx, image_name)
results[0].save(filename=output_path_onnx)

print(f"Inference result saved to: {output_path_onnx}")
```
