```python
import onnx
from ultralytics import YOLO
import os
```

```python
model = YOLO('train15/weights/best.pt')
model.export(format='onnx', imgsz=640)
```

```python
from onnxruntime.quantization import quantize_dynamic, QuantType

quantize_dynamic(
    model_input='train15/weights/best_640.onnx',
    model_output='quantized_models/best_640_quant_QUInt8.onnx',
    weight_type=QuantType.QUInt8  # or QuantType.QUInt8
)

print("Quantized model saved as best_640_quant.onnx")
```

```python
image_path = '/home/dll1305/Documents/ai-project/tf_models/object_detection/quantization/AI_Project/test/munich/'
image_name = "munich_000000_000019_leftImg8bit.png"

model_onnx = YOLO('/home/dll1305/Documents/ai-project/tf_models/object_detection/quantization/AI_Project/train15/weights/best.onnx', task="detect")
model_onnx_quant = YOLO('/home/dll1305/Documents/ai-project/tf_models/object_detection/quantization/AI_Project/quantized_models/best_quant.onnx', task="detect")

output_dir_onnx = '/home/dll1305/Documents/ai-project/tf_models/object_detection/quantization/AI_Project/output_onnx'
output_dir_onnx_quant = '/home/dll1305/Documents/ai-project/tf_models/object_detection/quantization/AI_Project/output_onnx_quant'

os.makedirs(output_dir_onnx, exist_ok=True)
os.makedirs(output_dir_onnx_quant, exist_ok=True)

results_onnx = model_onnx(image_path + image_name)
results_onnx_quant = model_onnx_quant(image_path + image_name)

output_path_onnx = os.path.join(output_dir_onnx, image_name)
output_path_onnx_quant = os.path.join(output_dir_onnx_quant, image_name)

results_onnx[0].save(filename=output_path_onnx)
results_onnx_quant[0].save(filename=output_path_onnx_quant)

print(f"Inference result saved to: {output_path_onnx}")
print(f"[Quant] Inference result saved to: {output_path_onnx_quant}")
```
